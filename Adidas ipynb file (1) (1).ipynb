# total Sales by region

```python
region_sales = df.groupby('Region')['Total Sales'].sum()

region_sales.plot(kind='bar', figsize=(8,5))
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

#Total sales by product

```python
product_sales = df.groupby('Product')['Total Sales'].sum()

product_sales.plot(kind='bar', figsize=(10,5))
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.xticks(rotation=75)
plt.show()
```

# units sold by sales method

```python
sales_method = df.groupby('Sales Method')['Units Sold'].sum()

sales_method.plot(kind='pie', autopct='%1.1f%%', figsize=(6,6))
plt.title('Units Sold by Sales Method')
plt.ylabel('')
plt.show()
```

```python
#Monthly Total Sales
```

```python
# Convert Invoice Date into datetime format
df['Invoice Date'] = pd.to_datetime(df['Invoice Date'])

# Extract Month
monthly_sales = df.groupby(df['Invoice Date'].dt.month)['Total Sales'].sum()

monthly_sales.plot(kind='line', marker='d', figsize=(8,5))
plt.title('Monthly Total Sales')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.grid(True)
plt.show()
```

# Histogram of price per unit

```python
df['Price per Unit'].plot(kind='hist', bins=10, figsize=(7,5))
plt.title('Distribution of Price per Unit')
plt.xlabel('Price per Unit')
plt.show()
```

```python
#Area chart by monthly profit
```

```python
monthly_profit = df.groupby(df['Invoice Date'].dt.month)['Operating Profit'].sum()

monthly_profit.plot(kind='area', figsize=(8,5))
plt.title('Monthly Operating Profit')
plt.xlabel('Month')
plt.ylabel('Operating Profit')
plt.show()
```

#Total 10 citites by total sales

```python
city_sales = df.groupby('City')['Total Sales'].sum().sort_values(ascending=False).head(10)

city_sales.plot(kind='bar', figsize=(10,5))
plt.title('Top 10 Cities by Total Sales')
plt.xlabel('City')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```
